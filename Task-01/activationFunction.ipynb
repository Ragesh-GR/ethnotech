```python
#task6
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import time
import pandas as pd

(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
x_train = x_train.reshape(-1, 28 * 28).astype('float32') / 255.0
x_test = x_test.reshape(-1, 28 * 28).astype('float32') / 255.0
y_train = keras.utils.to_categorical(y_train, 10)
y_test = keras.utils.to_categorical(y_test, 10)

print("MNIST data loaded and preprocessed.")
print(f"x_train shape: {x_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"x_test shape: {x_test.shape}")
print(f"y_test shape: {y_test.shape}")

def create_model(activation_function):
    model = keras.Sequential([
        keras.Input(shape=(784,)),
        layers.Dense(128, activation=activation_function),
        layers.Dense(64, activation=activation_function),
        layers.Dense(10, activation='softmax')
    ])
    return model
EPOCHS = 10
BATCH_SIZE = 32
OPTIMIZER = 'adam'
LOSS_FUNCTION = 'categorical_crossentropy'
METRICS = ['accuracy']
activation_functions = ['sigmoid', 'tanh', 'relu', 'leaky_relu']
results = []

for act_fn_name in activation_functions:
    print(f"\n--- Training with {act_fn_name.upper()} activation ---")

    if act_fn_name == 'leaky_relu':
        act_fn = layers.LeakyReLU(alpha=0.01)
    else:
        act_fn = act_fn_name

    model = create_model(act_fn)
    model.compile(optimizer=OPTIMIZER, loss=LOSS_FUNCTION, metrics=METRICS)
    start_time = time.time()
    history = model.fit(
        x_train, y_train,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        validation_split=0.1,
        verbose=0
    )
    training_time = time.time() - start_time
    test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)
    val_accuracy = history.history['val_accuracy'][-1]
    val_loss = history.history['val_loss'][-1]

    print(f"  Test Accuracy: {test_accuracy:.4f}")
    print(f"  Test Loss: {test_loss:.4f}")
    print(f"  Validation Accuracy: {val_accuracy:.4f}")
    print(f"  Validation Loss: {val_loss:.4f}")
    print(f"  Training Time: {training_time:.2f} seconds")

    results.append({
        'Activation Function': act_fn_name.upper(),
        'Test Accuracy': test_accuracy,
        'Test Loss': test_loss,
        'Validation Accuracy': val_accuracy,
        'Validation Loss': val_loss,
        'Training Time (s)': training_time
    })
results_df = pd.DataFrame(results)
print("\n--- Experiment Results ---")
display(results_df)
```

## Task 7: Real-world Case Study - Activation Function Recommendations

Here are recommendations for activation functions for various real-world applications, along with justifications:

### 1. House Price Prediction
*   **Recommendation**: **ReLU** or **Leaky ReLU** for hidden layers, **Linear** (no activation) for the output layer.
*   **Justification**: House price prediction is a regression task, meaning the output is a continuous numerical value (e.g., $300,000). For the output layer, a linear activation function is necessary to allow the model to predict any real number without bounding it. For hidden layers, ReLU or Leaky ReLU are generally preferred because they help mitigate the vanishing gradient problem, allowing deeper networks to learn effectively. They are computationally efficient and perform well in most deep learning tasks.

### 2. Cat vs Dog Classification
*   **Recommendation**: **ReLU** or **Leaky ReLU** for hidden layers, **Sigmoid** for the output layer (if binary classification) or **Softmax** (if multi-class classification, though here it's binary).
*   **Justification**: This is a binary classification task. For hidden layers, ReLU or Leaky ReLU are standard choices for their computational efficiency and ability to prevent vanishing gradients. For the output layer, Sigmoid is suitable as it outputs a probability between 0 and 1, representing the likelihood of being a 'cat' (or 'dog').

### 3. Sentiment Analysis
*   **Recommendation**: **ReLU** or **Leaky ReLU** for hidden layers, **Sigmoid** for the output layer (if binary sentiment: positive/negative) or **Softmax** (if multi-class sentiment: positive/negative/neutral).
*   **Justification**: Sentiment analysis is typically a classification task (e.g., positive/negative). Similar to image classification, ReLU or Leaky ReLU are excellent choices for hidden layers in deep learning models like LSTMs, GRUs, or Transformers. The output layer depends on the number of classes: Sigmoid for binary sentiment, Softmax for multiple sentiment categories (e.g., positive, negative, neutral).

### 4. Face Recognition
*   **Recommendation**: **ReLU** or **Leaky ReLU** for hidden layers.
*   **Justification**: Face recognition often involves learning complex features from image data using deep convolutional neural networks (CNNs). ReLU and its variants are almost universally used in CNNs due to their ability to prevent vanishing gradients, computational efficiency, and sparse activation, which can help in learning hierarchical features. The final output layer's activation depends on whether it's a classification (e.g., identify person) or embedding generation task.

### 5. Object Detection
*   **Recommendation**: **ReLU** or **Leaky ReLU** for hidden layers (in the backbone CNN), **Linear** for bounding box coordinates, **Sigmoid** for object presence confidence, and **Softmax** for class probabilities.
*   **Justification**: Object detection is a multi-task problem. The backbone feature extractor (often a CNN) heavily relies on ReLU or Leaky ReLU. The output layer for bounding box regression needs linear activation. For classifying if an object is present in a proposed bounding box, Sigmoid is used. For classifying *which* object it is (e.g., car, person, dog), Softmax is used. So, a combination of activations is typically employed in the output head.

### 6. Time Series Forecasting
*   **Recommendation**: **ReLU** or **Leaky ReLU** for hidden layers (e.g., in LSTMs, GRUs, or Transformers), **Linear** for the output layer.
*   **Justification**: Time series forecasting is a regression task where the model predicts future continuous values. Thus, a linear activation is required for the output layer. For the recurrent or convolutional layers that process the time series data, ReLU or Leaky ReLU are effective for their gradient-friendly properties and computational efficiency.

### 7. Medical Image Classification
*   **Recommendation**: **ReLU** or **Leaky ReLU** for hidden layers, **Sigmoid** (for binary classification, e.g., disease present/absent) or **Softmax** (for multi-class classification, e.g., different types of disease).
*   **Justification**: This is a critical classification task often using CNNs. ReLU and its variants are the go-to for hidden layers in CNN architectures due to their strong performance in image-related tasks, preventing vanishing gradients and promoting efficient learning of complex patterns in medical images. The choice for the output layer depends on the specific classification problem.

### 8. Fraud Detection
*   **Recommendation**: **ReLU** or **Leaky ReLU** for hidden layers, **Sigmoid** for the output layer.
*   **Justification**: Fraud detection is typically a binary classification problem (fraudulent/non-fraudulent). For tabular or sequence-based data often used in fraud detection, deep neural networks with ReLU or Leaky ReLU in hidden layers are effective for learning complex, non-linear patterns. A Sigmoid activation in the output layer provides a probability score indicating the likelihood of a transaction being fraudulent, which is useful for risk assessment.

```python
#task 10
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.activations import gelu
import numpy as np
import time
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

print("Loading and preprocessing MNIST data...")
(x_train_mnist, y_train_mnist), (x_test_mnist, y_test_mnist) = keras.datasets.mnist.load_data()

x_train_mnist = x_train_mnist.reshape(-1, 28 * 28).astype('float32') / 255.0
x_test_mnist = x_test_mnist.reshape(-1, 28 * 28).astype('float32') / 255.0
y_train_mnist = keras.utils.to_categorical(y_train_mnist, 10)
y_test_mnist = keras.utils.to_categorical(y_test_mnist, 10)

print(f"MNIST x_train shape: {x_train_mnist.shape}, y_train shape: {y_train_mnist.shape}")

print("Loading and preprocessing Fashion-MNIST data...")
(x_train_fmnist, y_train_fmnist), (x_test_fmnist, y_test_fmnist) = keras.datasets.fashion_mnist.load_data()

x_train_fmnist = x_train_fmnist.reshape(-1, 28 * 28).astype('float32') / 255.0
x_test_fmnist = x_test_fmnist.reshape(-1, 28 * 28).astype('float32') / 255.0

y_train_fmnist = keras.utils.to_categorical(y_train_fmnist, 10)
y_test_fmnist = keras.utils.to_categorical(y_test_fmnist, 10)

print(f"Fashion-MNIST x_train shape: {x_train_fmnist.shape}, y_train shape: {y_train_fmnist.shape}")
def create_mlp_model(activation_function):
    model = keras.Sequential([
        keras.Input(shape=(784,)),
        layers.Dense(128, activation=activation_function),
        layers.Dense(64, activation=activation_function),
        layers.Dense(10, activation='softmax')
    ])
    return model

EPOCHS = 10
BATCH_SIZE = 32
OPTIMIZER = 'adam'
LOSS_FUNCTION = 'categorical_crossentropy'
METRICS = ['accuracy']
activation_functions = {
    'Sigmoid': 'sigmoid',
    'Tanh': 'tanh',
    'ReLU': 'relu',
    'Leaky ReLU': layers.LeakyReLU(alpha=0.01),
    'ELU': layers.ELU(alpha=1.0),
    'Swish': 'swish',
    'GELU': gelu
}

all_results = []
all_histories = {}
datasets = {
    'MNIST': {
        'x_train': x_train_mnist, 'y_train': y_train_mnist,
        'x_test': x_test_mnist, 'y_test': y_test_mnist
    },
    'Fashion-MNIST': {
        'x_train': x_train_fmnist, 'y_train': y_train_fmnist,
        'x_test': x_test_fmnist, 'y_test': y_test_fmnist
    }
}

for dataset_name, data in datasets.items():
    print(f"\n{'='*50}")
    print(f"Starting experiments for {dataset_name}")
    print(f"{'='*50}")

    dataset_history = {}

    for act_fn_name, act_fn_impl in activation_functions.items():
        print(f"\n--- Training with {act_fn_name} activation on {dataset_name} ---")
        model = create_mlp_model(act_fn_impl)
        model.compile(optimizer=OPTIMIZER, loss=LOSS_FUNCTION, metrics=METRICS)
        start_time = time.time()
        history = model.fit(
            data['x_train'], data['y_train'],
            epochs=EPOCHS,
            batch_size=BATCH_SIZE,
            validation_split=0.1,
            verbose=0
        )
        training_time = time.time() - start_time
        test_loss, test_accuracy = model.evaluate(data['x_test'], data['y_test'], verbose=0)
        y_pred_probs = model.predict(data['x_test'], verbose=0)
        y_pred_classes = np.argmax(y_pred_probs, axis=1)
        y_true_classes = np.argmax(data['y_test'], axis=1)
        precision = precision_score(y_true_classes, y_pred_classes, average='weighted', zero_division=0)
        recall = recall_score(y_true_classes, y_pred_classes, average='weighted', zero_division=0)
        f1 = f1_score(y_true_classes, y_pred_classes, average='weighted', zero_division=0)
        cm = confusion_matrix(y_true_classes, y_pred_classes)
        val_accuracy = history.history['val_accuracy'][-1]
        val_loss = history.history['val_loss'][-1]
        train_accuracy = history.history['accuracy'][-1]
        train_loss = history.history['loss'][-1]

        print(f"  Training Accuracy: {train_accuracy:.4f}")
        print(f"  Training Loss: {train_loss:.4f}")
        print(f"  Validation Accuracy: {val_accuracy:.4f}")
        print(f"  Validation Loss: {val_loss:.4f}")
        print(f"  Test Accuracy: {test_accuracy:.4f}")
        print(f"  Test Loss: {test_loss:.4f}")
        print(f"  Precision: {precision:.4f}")
        print(f"  Recall: {recall:.4f}")
        print(f"  F1-Score: {f1:.4f}")
        print(f"  Training Time: {training_time:.2f} seconds")

        all_results.append({
            'Dataset': dataset_name,
            'Activation Function': act_fn_name,
            'Training Accuracy': train_accuracy,
            'Training Loss': train_loss,
            'Validation Accuracy': val_accuracy,
            'Validation Loss': val_loss,
            'Test Accuracy': test_accuracy,
            'Test Loss': test_loss,
            'Precision': precision,
            'Recall': recall,
            'F1-Score': f1,
            'Training Time (s)': training_time,
            'Confusion Matrix': cm
        })
        dataset_history[act_fn_name] = history.history
    all_histories[dataset_name] = dataset_history

results_df = pd.DataFrame(all_results)
print("\n--- Combined Experiment Results ---")
display(results_df)

for dataset_name, histories in all_histories.items():
    plt.figure(figsize=(18, 10))
    plt.suptitle(f'Loss Curves for {dataset_name} Dataset', fontsize=16)

    num_activations = len(histories)
    cols = 3
    rows = (num_activations + cols - 1) // cols

    for i, (act_fn_name, history) in enumerate(histories.items()):
        plt.subplot(rows, cols, i + 1)
        plt.plot(history['loss'], label='Training Loss')
        plt.plot(history['val_loss'], label='Validation Loss')
        plt.title(f'{act_fn_name}')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.legend()
        plt.grid(True)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

    plt.figure(figsize=(18, 10))
    plt.suptitle(f'Confusion Matrices for {dataset_name} Dataset', fontsize=16)
    dataset_specific_results = [res for res in all_results if res['Dataset'] == dataset_name]
    num_activations = len(dataset_specific_results)
    cols = 3
    rows = (num_activations + cols - 1) // cols
    for i, res in enumerate(dataset_specific_results):
        cm = res['Confusion Matrix']
        act_fn_name = res['Activation Function']
        plt.subplot(rows, cols, i + 1)
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
        plt.title(f'{act_fn_name}')
        plt.xlabel('Predicted Label')
        plt.ylabel('True Label')
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

comparison_table_df = results_df.drop(columns=['Confusion Matrix'])
display(comparison_table_df)
```
