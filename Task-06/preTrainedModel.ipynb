```python
from keras.applications import VGG16
from tensorflow.keras.preprocessing import image
from keras.applications.vgg16 import preprocess_input, decode_predictions
import numpy as np
```

```python
model = VGG16(weights='imagenet', include_top=True, input_shape = (224, 224, 3))
```

```python
# Loading an image and processee it for the model

img1_path = 'birds.jpg'
img2_path = 'cat.jpg'
img3_path = 'dog.jpg'
img4_path = 'chocolate_cake.jpeg'
img5_path = 'box.webp'
```

```python
image_paths = [img1_path, img2_path, img3_path, img4_path, img5_path]
processed_images = []

for img_path in image_paths:
    img = image.load_img(img_path, target_size=(224, 224))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) # Add batch dimension
    img_array = preprocess_input(img_array)
    processed_images.append(img_array)

# Concatenate all processed images into a single NumPy array
# This creates a batch of images ready for the model
processed_images_batch = np.vstack(processed_images)

print(f"Shape of the preprocessed image batch: {processed_images_batch.shape}")
```

```python
# Use the VGG16 model to predict classes for the preprocessed images
predictions = model.predict(processed_images_batch)

# Decode the predictions to get human-readable labels and confidence scores
# For each image, get the top 3 predictions
decoded_predictions = decode_predictions(predictions, top=3)

for i, img_predictions in enumerate(decoded_predictions):
    print(f"Predictions for image {i+1} (Path: {image_paths[i]}):")
    for class_id, class_name, confidence in img_predictions:
        print(f"  - {class_name} ({confidence:.2f})")
    print("\n")
```
